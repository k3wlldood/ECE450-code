```python
""" find_n """

import matplotlib.pyplot as plt
import numpy as np
from scipy import signal as sig
from math import pi, exp, cos, sin, log, log10, sqrt
```

```python
""" Low Pass Filter """

wp = .8
Hp = .9
ws = 3.33
Hs = .15

wc = 1.2 #guess this

wp = wp/wc
ws = ws/wc

Hp2 = (Hp)**2

np = log10(1/Hp2 - 1)/(2*log10(wp))
print('np =', np)

Hs2 = (Hs)**2

ns = log10(1/Hs2 - 1)/(2*log10(ws))
print('ns =', ns)
```

```python
""" High Pass Filter """

wp = 80
Hp = .8
ws = 20
Hs = .2

wc = 65 #guess this

#Scale down and flip to LP

wp = 1/(wp/wc)
ws = 1/(ws/wc)

Hp2 = (Hp)**2

np = log10(1/Hp2 - 1)/(2*log10(wp))
print('np =', np)

Hs2 = (Hs)**2

ns = log10(1/Hs2 - 1)/(2*log10(ws))
print('ns =', ns)
```
