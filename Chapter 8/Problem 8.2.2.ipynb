```python
""" Problem 8.2.2 """

import matplotlib.pyplot as plt
import numpy as np
from scipy import signal as sig
from math import pi, exp, cos, sin, log, log10, sqrt
from control import margin
```

```python
a = 50e3
n2 = [1,0,0]
d2 = [1,1.41*a,a**2]

num = n2
den = d2

w = np.linspace(10000,1000000,1000)
system = sig.lti(num,den)
w, Hmag, Hphase = sig.bode(system,w)
gm, pm, wg, wp = margin(Hmag,Hphase,w)

plt.subplot(211)
plt.semilogx(w,Hmag,'k')
plt.axis([ 10000, 1e6, -20, 10])
plt.ylabel('|H| dB',size = 12)
plt.grid(which='both')
plt.xticks([1e4,50e3,1e5,1e6])
plt.yticks([-20,-10,-3,0,10])
plt.xlabel('$\omega$ (rad/s)')
plt.text(40e3,-24.5,'$\\mathdefault{50\\times10^{3}}$')
```
