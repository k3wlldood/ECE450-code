```python
""" Test_FFT """

import matplotlib.pyplot as plt
import numpy as np
from scipy import signal as sig
from math import pi, exp, cos, sin, log, sqrt

NN = 100
N2 = int(NN/2)
x = np.zeros(NN)
y = np.zeros(NN)

dt = .1
TT = np.linspace(0,dt*(NN-1),NN)
DF = 1/(dt*NN)
FF = np.linspace(0,DF*(NN-1),NN)

""" Create the time-domain input """

freq1 = 8*pi/NN
freq2 = 3*pi/NN
freq3 = 4*pi/NN

x = np.sin(freq1*TT) + 1*np.sin(freq2*TT) + 1*np.sin(freq3*TT)

plt.subplot(311)
plt.plot(x,'k')
#plt.axis([0,NN*dt,-2.5,2.5])
#plt.text(5,-15,'$\phi$ = {}'.format(round(14,3)),fontsize=12)
plt.title('Test_FFT')
plt.ylabel(' x[k]')
plt.xlabel('T (sec)')
plt.grid()

"""Take the fft of x"""
X = (1/NN)*np.fft.fft(x)

plt.subplot(312)
plt.plot(abs(X),'k',label='X')
#plt.legend(loc='upper right')
plt.ylabel('X(w)')
plt.xlabel('Freq (Hz)')
plt.grid()
#plt.savefig('f.png')
# plt.axis([0,200,0,1.1])
# plt.xticks([20,80,150])

""" Take the ifft of X"""
y = NN*np.fft.ifft(X)

plt.subplot(313)
plt.plot(y,'k',label='X')
plt.ylabel('y')
plt.grid()
plt.show()
```
