```python
""" time_domain.py """

import matplotlib.pyplot as plt
import numpy as np
from scipy import signal as sig
from math import pi, exp, cos, sin, log, sqrt
from control import margin
from control import tf
from math import pi, sqrt, exp, cos,sin
from cmath import exp

""" WILL NOT RUN ON ITS OWN """

""" PLACE CODE FOR ZBODE HERE """

dt = .01
MM = 5000
x = np.zeros(MM)
y = np.zeros(MM)
w1 = 0.5
w2 = 5

for k in range(1,MM):
    x[k] = 1*sin(w1*dt*k) + 0.5* sin(w2*dt*k)
    y[k] = .905*y[k-1]+.095*x[k]
    a1 = .001
    y[k] = 1.89*y[k-1]-.891*y[k-2]+a1*x[k-1]


plt.subplot(211)
plt.plot(x,'k')
plt.ylabel('x(t)')
plt.title('Time_dom')
plt.grid()

plt.subplot(212)
plt.plot(y,'k')
plt.ylabel('y(t)')
plt.xlabel('Time (ms)')
plt.axis([0,MM,-1.2,1.2])
plt.yticks([-1,-.7,-.1,.1,.7,1])
plt.grid()
plt.tight_layout()
plt.show()
```
